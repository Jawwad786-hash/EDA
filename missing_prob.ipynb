```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.arange(1, 21).reshape(5, 4),
                  columns=['A', 'B', 'C', 'D'])

df.iloc[0, 0] = np.nan
df.iloc[1, 1] = np.nan
df.iloc[2, 2] = np.nan
df.iloc[3, 3] = np.nan

print(df)
print(df.isnull().sum())
```
