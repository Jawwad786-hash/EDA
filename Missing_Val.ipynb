```python
import pandas as pd
import numpy as np
data=pd.Series([1,np.NAN,3,None,5],index=['a','b','c','d','e'])
print(data)
print(data.isnull())
print(data.notnull())
```

```python
data=pd.Series([1,np.NAN,3,None,5],index=['a','b','c','d','e'])
data1=data.dropna()
print(data)
print(data1)
```

```python
data3=pd.DataFrame([[1,np.nan,2],[2,3,5],[np.nan,4,6]])
print(data3)
print(data3.dropna())
print(data3.dropna(axis='columns'))
print(data3.dropna(axis='columns',how='all'))
print(data3.dropna(axis='rows',thresh=3))

```

```python
data4=pd.DataFrame([[1,np.nan,2],[2,3,5],[np.nan,4,6]])
print(data4)
print(data4.fillna(1))
print(data4.fillna(method='ffill'))
print(data4.fillna(method='bfill'))
```
